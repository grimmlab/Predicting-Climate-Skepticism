```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('ggplot')

final_results = pd.read_csv(final_results)
best_rmse = final_results['test_refitting_period_complete_rmse'].iloc[0]
RMSE = round(best_rmse)

pred = final_results['y_pred_test_refitting_period_complete'].dropna().values
true = final_results['y_true_test'].dropna().values

x = list(range(1, len(pred) + 1))

ax = plt.axes((0.1, 0.1, 1.0, 0.8))
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['top'].set_visible(False)

ax.plot(x, pred, marker='o', markersize=4, markerfacecolor='#c1272d',
        markeredgecolor='black', markeredgewidth=1, color='#c1272d',
        linestyle='--', linewidth=2)
ax.plot(x, true, marker='o', markersize=4, markerfacecolor='#0000a7',
        markeredgecolor='black', markeredgewidth=1, color='#0000a7',
        linestyle='--', linewidth=2)

ax.grid(True, which='major', axis='both', alpha=0.3)

plt.xlabel("Time Step")
plt.ylabel("traffic_volume")

if len(x) > 25:
    ax.xaxis.set_major_locator(plt.MaxNLocator(25))
else:
    plt.xticks(x)

ax.legend(['Forecasts(complete) with ard (RMSE: ' + str(RMSE) + ')', "True Data"], frameon=True)

fig1 = plt.gcf()
plt.show()
plt.draw()

fig1.savefig('Fig3_ard_dataset_full_best_refitting_cycle_complete.svg', format='svg', bbox_inches='tight')
```
