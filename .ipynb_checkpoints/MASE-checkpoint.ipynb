```python
def MASE(mae, Actual, Predicted, last_trainval_value):
    mae_native_sum = 0
    for i in range (0, len(Actual)):
        if i == 0:
            naive = last_trainval_value
        else:
            naive = Actual[i - 1]
        mae_native_sum += abs(Actual[i] - naive)
    mae_native = mae_native_sum/(len(Actual))
    return mae/mae_native
```

```python
import os
import pandas as pd
pd.set_option('display.max_columns', 500)
from sklearn.metrics import mean_absolute_percentage_error, mean_absolute_error
import numpy as np
import traceback

def get_best_models(
    rootdir='/bit_storage/Workspace/Josef/01_Schreibtisch/02_ForeTiSHortiCo/01_ForeTiS-HortiCo/02_Experiments/General Model/results'
):

    refitting_cycles_list = [
        'complete', 0, 1, 2, 4, 8
    ]

    models_list = [
        "arima", "xgboost"
    ]

    retailer_list = ["schachtschneider", "keller", "blumenzentrale", "heusser", "peters", "offenloch"]


    metrics_list = ['MAE', 'MAPE', 'MASE']

    d = dict()
    d_top = dict()

    for model in os.listdir(rootdir):
        for retailer_featureset in os.listdir(rootdir + '/' + model):
            if os.path.isdir(os.path.join(
                    rootdir, model, retailer_featureset)) and retailer_featureset[0] != '.' and retailer_featureset!= "Archiv":
                if model == "arima":
                    for featureset in os.listdir(rootdir + '/' + model +
                                             '/' + retailer_featureset):
                        results = pd.read_csv(
                            rootdir + '/' + model + '/' +
                            retailer_featureset + '/' + featureset +
                            '/final_model_test_results.csv')
                        last_trainval_value = results["y_true_retrain"].dropna().iloc[-1]
                        for i, refitting_cycle in enumerate(refitting_cycles_list):
                            y_true = results["y_true_test"].dropna()
                            try:
                                y_pred = results["y_pred_test_refitting_period_" + str(refitting_cycle)].dropna()
                            except KeyError as exc:
                                print(retailer_featureset)
                                print(model)
                                print(featureset)
                                print(traceback.format_exc())
                                print(exc)
                                continue
                            if len(y_pred) == (len(y_true) - 1):
                                y_true = y_true[:-1]
                            mae = mean_absolute_error(y_true, y_pred)
                            mape = mean_absolute_percentage_error(y_true, y_pred)
                            mase = MASE(mae, y_true, y_pred, last_trainval_value)
                            results.insert(results.columns.get_loc('test_refitting_period_' + str(refitting_cycle) + '_r2_score'), column='test_refitting_period_' + str(refitting_cycle) + '_mae', value=np.nan)
                            results.at[0, 'test_refitting_period_' + str(refitting_cycle) + '_mae'] = mae
                            results.insert(results.columns.get_loc('test_refitting_period_' + str(refitting_cycle) + '_r2_score'), column='test_refitting_period_' + str(refitting_cycle) + '_mape', value=np.nan)
                            results.at[0, 'test_refitting_period_' + str(refitting_cycle) + '_mape'] = mape
                            results.insert(results.columns.get_loc('test_refitting_period_' + str(refitting_cycle) + '_r2_score'), column='test_refitting_period_' + str(refitting_cycle) + '_mase', value=np.nan)
                            results.at[0, 'test_refitting_period_' + str(refitting_cycle) + '_mase'] = mase
                        results.to_csv(model + '_' +
                            retailer_featureset + '_' + featureset +
                            '_final_model_test_results_metrics.csv')  
                else:
                    for retailer_count in retailer_list:
                        for featureset in os.listdir(rootdir + '/' + model +
                                             '/' + retailer_featureset +
                                             '/' + retailer_count):
                            results = pd.read_csv(
                            rootdir + '/' + model + '/' +
                            retailer_featureset + '/' + retailer_count + '/' + featureset +
                            '/final_model_test_results.csv')
                            last_trainval_value = results["y_true_retrain"].dropna().iloc[-1]
                            for i, refitting_cycle in enumerate(refitting_cycles_list):
                                y_true = results["y_true_test"].dropna()
                                try:
                                    y_pred = results["y_pred_test_refitting_period_" + str(refitting_cycle)].dropna()
                                except KeyError as exc:
                                    print(retailer_featureset)
                                    print(model)
                                    print(featureset)
                                    print(retailer_count)
                                    print(traceback.format_exc())
                                    print(exc)
                                    continue
                                if len(y_pred) == (len(y_true) - 1):
                                    y_true = y_true[:-1]
                                mae = mean_absolute_error(y_true, y_pred)
                                mape = mean_absolute_percentage_error(y_true, y_pred)
                                mase = MASE(mae, y_true, y_pred, last_trainval_value)
                                results.insert(results.columns.get_loc('test_refitting_period_' + str(refitting_cycle) + '_r2_score'), column='test_refitting_period_' + str(refitting_cycle) + '_mae', value=np.nan)
                                results.at[0, 'test_refitting_period_' + str(refitting_cycle) + '_mae'] = mae
                                results.insert(results.columns.get_loc('test_refitting_period_' + str(refitting_cycle) + '_r2_score'), column='test_refitting_period_' + str(refitting_cycle) + '_mape', value=np.nan)
                                results.at[0, 'test_refitting_period_' + str(refitting_cycle) + '_mape'] = mape
                                results.insert(results.columns.get_loc('test_refitting_period_' + str(refitting_cycle) + '_r2_score'), column='test_refitting_period_' + str(refitting_cycle) + '_mase', value=np.nan)
                                results.at[0, 'test_refitting_period_' + str(refitting_cycle) + '_mase'] = mase
                            results.to_csv(model + '_' +
                                retailer_featureset + '_' + retailer_count + '_' + featureset +
                                '_final_model_test_results_metrics.csv')
```
