```python
import sklearn.metrics
import matplotlib.pyplot as plt
import pandas as pd
test = pd.read_csv('../Utils/Results/xgb/climate_knowledge/2024-12-10_11-31-35/test.csv')
predictions = pd.read_csv("../Utils/Results/xgb/climate_knowledge/2024-12-10_11-31-35/predictions.csv")
disp = sklearn.metrics.ConfusionMatrixDisplay.from_predictions(
            test.iloc[:,125], predictions, cmap=plt.cm.Blues)
print(disp.confusion_matrix)
```
