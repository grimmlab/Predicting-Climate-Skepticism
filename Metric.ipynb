```python
import sklearn
import matplotlib.pyplot as plt
import pandas as pd
test = pd.read_csv('../Utils/Results/xgb/wo_personal_wo_climate_state_belief/2025-01-30_17-15-01/test.csv')
predictions = pd.read_csv("../Utils/Results/xgb/wo_personal_wo_climate_state_belief/2025-01-30_17-15-01/predictions.csv")
disp = sklearn.metrics.ConfusionMatrixDisplay.from_predictions(
            test.iloc[:,165], predictions, cmap=plt.cm.Blues)
print(disp.confusion_matrix)
print(sklearn.metrics.matthews_corrcoef(predictions, test.iloc[:,165]))
```
