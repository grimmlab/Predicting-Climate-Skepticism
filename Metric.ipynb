```python
import os, glob
import pandas as pd
import sklearn.metrics

def metric():
    for dependent_variable in os.listdir("../Utils/Results/results"):
        if dependent_variable != ".DS_Store":
            for experiment in os.listdir(f"../Utils/Results/results/{dependent_variable}"):
                if experiment != ".DS_Store":
                    latest_file = max(glob.glob(os.path.join(f"../Utils/Results/results/{dependent_variable}/{experiment}/xgb", '*/')), key=os.path.getmtime)
                    predictions = pd.read_csv(f'{latest_file}/predictions.csv', header=None)
                    test = pd.read_csv(f'{latest_file}/test.csv')
                    if dependent_variable == "climatedeniers":
                        try:
                            score = sklearn.metrics.matthews_corrcoef(predictions, test["climatedeniers"])
                        except KeyError:
                            continue
                    else:
                        try:
                            score = sklearn.metrics.r2_score(predictions, test[dependent_variable])
                        except KeyError:
                            continue
                    try:
                        with open(f'{latest_file}/score_{experiment}.txt', 'w') as f:
                            f.write(str(score))
                    except ValueError:
                        continue
```

```python
metric()
```
