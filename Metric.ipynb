```python
import sklearn.metrics
import matplotlib.pyplot as plt
import pandas as pd
test = pd.read_csv('../Utils/Results/xgb/2024-11-07_13-17-58/test.csv')
predictions = pd.read_csv("../Utils/Results/xgb/2024-11-07_13-17-58/predictions.csv")
disp = sklearn.metrics.ConfusionMatrixDisplay.from_predictions(
            test.iloc[:,-1], predictions, cmap=plt.cm.Blues)
print(disp.confusion_matrix)
```
