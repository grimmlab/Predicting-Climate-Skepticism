```python
import joblib
import shap
import matplotlib.pyplot as plt
shap_values = joblib.load(open("../Utils/Results/xgb/climate_policies_actions_east_germany_east_germany/2025-03-11_07-13-17/shapvalues.sav", 'rb'))
shap.plots.beeswarm(shap_values, max_display=25, color="viridis", show=False, order=shap_values.abs.max(0), color_bar_label="Value of analyzed product characteristic")
fig1 = plt.gcf()
plt.show()
plt.draw()
fig1.savefig('plots/beeplot.svg', format='svg', bbox_inches='tight')
```
